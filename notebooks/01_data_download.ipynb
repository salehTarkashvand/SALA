## 7. Price overview chart

Plot the closing price of each asset on its own subplot to visually inspect
the data. This is a quick sanity check before moving on to modeling.

**What to look for:**
- No large gaps (missing days)
- No sudden jumps that look like data errors
- The overall trend makes sense (e.g., BTC should be volatile, gold calmer)

**Output:** `results/figures/01_price_overview.png`

```python
fig, axes = plt.subplots(len(data), 1, figsize=(12, 2.8 * len(data)), sharex=True)

if len(data) == 1:
    axes = [axes]

for ax, (ticker, df) in zip(axes, data.items()):
    ax.plot(df.index, df["Close"], linewidth=1.2)
    ax.set_title(f"{ticker} — {TICKERS[ticker]}", fontsize=11)
    ax.set_ylabel("Price (USD)")
    ax.grid(True, alpha=0.3)
    ax.tick_params(labelsize=9)

plt.xlabel("Date")
plt.tight_layout()

FIG_PATH = PROJECT_ROOT / "results" / "figures" / "01_price_overview.png"
FIG_PATH.parent.mkdir(parents=True, exist_ok=True)
plt.savefig(FIG_PATH, dpi=120, bbox_inches="tight")
plt.show()

print(f"Figure saved: {FIG_PATH.name}")
```

## 8. Daily returns

Compute the daily return for each asset:

$$
r_t = \frac{P_t - P_{t-1}}{P_{t-1}}
$$

This is the core quantity used in later stages:
- **Anomaly detector** flags unusual returns.
- **Learning automaton** receives returns as the reinforcement signal.

**What to look for in the table:**
- Mean should be close to 0 for most assets (slight positive drift is normal).
- Standard deviation is a measure of volatility — BTC/ETH should have higher
  std than gold or the S&P 500.
- Min/Max show the worst and best single days — useful for spotting crashes.

```python
# Compute daily returns for each asset
print(f"{'Ticker':<10} {'Mean':>10} {'Std':>10} {'Min':>10} {'Max':>10}")
print("-" * 55)
for ticker, df in data.items():
    r = df["Close"].pct_change().dropna()
    print(f"{ticker:<10} {r.mean():>10.6f} {r.std():>10.6f} {r.min():>10.6f} {r.max():>10.6f}")
```
